```python
# Find all words in fname that contain search_str, ignoring case
def print_words_with_str(fname, search_str):
    # Open fname (for reading) and iterate through each line in it
    with open(fname) as infile:
        for line in infile:
            # Split line into a list based on white space
            line = line.split()
            # The frequency is at index 0, the word is at index 1
            word = line[1]
            # Check if search_str is a substring of word; apply case folding
            # before cheking
            if search_str.lower() in word.lower():
                print(word)
```

```python
# Words in dracula containing the substring cat
print_words_with_str('dracula_freqs.txt', 'cat')
```

```python
# Return a boolean indicating whether s is a palindrome, optionally ignoring case
def is_palindrome(s, apply_case_folding=True):
    if apply_case_folding:
        s = s.lower()
    # Here are two different ways to reverse a string
    rev_s = ''.join(reversed(s))
    #rev_s = s[::-1]
    return s == rev_s
```

```python
# Find palindromes in fname
def find_palindromes(fname):
    palindromes = set()
    
    with open(fname) as infile:
        for line in infile:
            line = line.split()
            word = line[1]
            # The length check just avoids finding boring palindromes,
            # like "a" and "aa"
            if is_palindrome(word) and len(word) > 2:
                palindromes.add(word)
        
    return sorted(palindromes, key=lambda x : len(x), reverse=True)
```

```python
# Palindromes in dracula
for p in find_palindromes('dracula_freqs.txt'):
    print(p)
```

```python
# Find the top-n most frequent words
def top_n(fname, n=5):
    # freqs is a dictionary (associative array)
    freqs = {}

    # Open fname (for reading) and iterate through each line in it
    with open(fname) as infile:
        for line in infile:
            num,word = line.split()
            num = int(num)
            freqs[word] = num

    # Find the top-n most frequent words
    top_n_words = sorted(freqs, key=lambda x : freqs[x], reverse=True)[:n]

    return top_n_words
```

```python
# Top-5 most frequent words in dracula
for w in top_n('dracula_freqs.txt'):
    print(w)
```

```python
# Top-10 most frequent words in alice's adventures in wonderland
for w in top_n('alice_freqs.txt', 10):
    print(w)
```

```python
# Return the average length of words that occur only once (also
# known as hapax legomena, or just hapaxes) in fname
def ave_hapax_len(fname):
    with open(fname) as infile:
        hapax_lens = []
        for line in infile:
            freq,word = line.split()
            freq = int(freq)
            if freq == 1:
                hapax_lens.append(len(word))

    return sum(hapax_lens) / len(hapax_lens)
```

```python
# Average hapax length for dracula
print(ave_hapax_len('dracula_freqs.txt'))
```

```python
# Same as above but using list comprehensions
def ave_hapax_len_2(fname):
    with open(fname) as infile:
        word_freqs = [x.split() for x in infile]
        hapax_lens = [len(x[1]) for x in word_freqs if int(x[0]) == 1]
        return (sum(hapax_lens) / float(len(hapax_lens)))
```

```python
print(ave_hapax_len_2('dracula_freqs.txt'))
```

```python
# An example class in Python for a vocabulary.
class Vocab:
    def __init__(self, fname):
        self.freqs = {}

        with open(fname) as infile:
            for line in infile:
                freq,word = line.split()
                self.freqs[word] = int(freq)

    def freq_for_word(self, word):
        return self.freqs.get(word, 0)

    def num_types(self):
        return len(self.freqs)

    def num_instances(self):
        # Note that this is inefficient because the sum is calculated
        # every time this method is called. (This starts to matter if
        # the dictionary is very large, and this method is called
        # often.) This calculation could be done once in the
        # constructor instead, and this method could simply return the
        # calculated value.
        return sum(self.freqs.values())
```

```python
v = Vocab('alice_freqs.txt')
print(v.freq_for_word('cat'))
print(v.freq_for_word('broccoli'))
print(v.num_types())
print(v.num_instances())
```
